```python
from unsloth.chat_templates import get_chat_template
from datasets import load_dataset

tokenizer = get_chat_template(tokenizer, chat_template="gemma-4-thinking")

dataset = load_dataset("json", data_files="xspirit_dataset.jsonl", split="train")

def formatting_prompts_func(examples):
    convos = examples["messages"]
    texts = [
        tokenizer.apply_chat_template(c, tokenize=False, add_generation_prompt=False).removeprefix('<bos>')
        for c in convos
    ]
    return {"text": texts}

dataset = dataset.map(formatting_prompts_func, batched=True)
```

```python
from trl import SFTTrainer, SFTConfig
from unsloth.chat_templates import train_on_responses_only
from google.colab import drive

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    args=SFTConfig(
        dataset_text_field="text",
        per_device_train_batch_size=1,
        gradient_accumulation_steps=4,
        max_steps=60,
        learning_rate=2e-4,
        output_dir="outputs",
    ),
)

trainer = train_on_responses_only(
    trainer,
    instruction_part="<|turn>user\n",
    response_part="<|turn>model\n",
)

trainer.train()

drive.mount('/content/drive')
model.save_pretrained("/content/drive/MyDrive/xspirit_e2b_abliterated")
tokenizer.save_pretrained("/content/drive/MyDrive/xspirit_e2b_abliterated")
```
